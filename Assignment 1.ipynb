Part A: Theory

1. What is Object-Oriented Programming (OOP)?

Ans - Object-Oriented Programming (OOP) is a programming approach that organizes software around objects, which combine data (attributes) and behavior (methods). It commonly uses concepts such as classes, objects, encapsulation, abstraction, inheritance, and polymorphism. OOP can make programs easier to organize, reuse, test, and maintain.


2. What is a class in OOP?

Ans - A class is a blueprint for creating objects. It defines the attributes (data) and methods (behavior) that objects of that type can have. For example, a Car class might define attributes such as color and methods such as start().


3. What is an object in OOP?

Ans - An object is an instance of a class. It has its own state, represented by its attributes, and can perform behavior through the class's methods. For example, my_car = Car() creates an object if Car is a defined class


4. What is the difference between abstraction and encapsulation?

Ans - Abstraction exposes essential features while hiding unnecessary implementation details; it focuses on what an object does. Encapsulation bundles data and methods together and controls how internal state is accessed or changed; it focuses on protecting and managing an object's internals. In Python, abstract base classes can support abstraction, while naming conventions and properties can support encapsulation.


5. What are dunder methods in Python?

Ans - Dunder methods (“double underscore” methods) are special methods whose names begin and end with double underscores, such as __init__, __str__, and __add__. Python calls them for built-in language operations—for example, __str__ supports string conversion and __add__ can define behavior for the + operator.


6. Explain the concept of inheritance in OOP.

Ans - Inheritance allows a new class (child or subclass) to reuse and extend the attributes and methods of an existing class (parent or superclass). It supports code reuse and class hierarchies. A child class may add new behavior or override inherited methods.


7. What is polymorphism in OOP?

Ans - Polymorphism means “many forms”: the same interface or method call can produce different behavior depending on the object. For example, different classes may each implement speak(), and calling speak() on each object invokes its own implementation.


8. How is encapsulation achieved in Python?

Ans - Python uses conventions and language features to manage access to object data. A single leading underscore (for example, _value) signals that an attribute is intended for internal use. A double leading underscore (for example, __balance) triggers name mangling, which helps avoid accidental access or name clashes but is not strict privacy.

9. What is a constructor in Python?

Ans - A constructor is commonly described as the method that initializes a new object. In typical Python classes, __init__(self, ...) initializes the object's attributes after the instance has been created. Technically, __new__ creates the instance, while __init__ initializes it.


10. What are class and static methods in Python?

Ans - A class method is defined with classmethod and receives the class as its first argument, conventionally named cls. It can access or modify class-level state and is often used for alternative constructors. A static method is defined with staticmethod and receives no automatic self or cls argument; it is used for a related utility that does not need instance or class state.


11. What is method overloading in Python?

Ans - Method overloading usually means defining multiple methods with the same name but different parameter lists. Python does not support traditional signature-based overloading in the same way as languages such as Java: a later definition replaces an earlier one. Similar behavior can be designed with default arguments, *args/**kwargs, or dispatch tools such as functools.singledispatch for functions.


12. What is method overriding in OOP?

Ans - Method overriding occurs when a subclass defines a method with the same name as a method in its parent class, providing a specialized implementation. When the method is called on a subclass object, Python normally uses the subclass version


13. What is a property decorator in Python?

Ans - The property decorator lets a method be accessed like an attribute. It is useful for calculated values and for controlling attribute access. A property can also define a setter with name.setter and a deleter with name.deleter.


14. Why is polymorphism important in OOP?

Ans - Polymorphism lets code work with different object types through a common interface. It reduces the need for type-specific branching, makes code easier to extend, and supports flexible designs—for example, a loop can call draw() on different shape objects without needing to know each concrete class.


15. What is an abstract class in Python?

Ans - An abstract class is a class intended to define a common interface and possibly shared behavior, while preventing direct instantiation when abstract methods remain unimplemented. Python provides abc.ABC and abstractmethod to implement abstract classes.


16. What are the advantages of OOP?

Ans - Key advantages include modular organization, code reuse through inheritance and composition, encapsulation of state, abstraction of implementation details, polymorphic and extensible designs, and easier maintenance and testing when classes have clear responsibilities. OOP is not always the best fit for every problem; design quality matters.


17. What is the difference between a class variable and an instance variable?

Ans - A class variable is defined on the class and is shared by instances unless shadowed. An instance variable belongs to a particular object and usually is assigned through self inside __init__. Use class variables for shared information (such as a count) and instance variables for object-specific information (such as each person's name).


18. What is multiple inheritance in Python?

Ans - Multiple inheritance means a class inherits from more than one parent class, for example class Child(ParentA, ParentB): .... Python uses method resolution order (MRO) to determine which inherited implementation is selected. Multiple inheritance can be useful but should be designed carefully to avoid confusing hierarchies.


19. Explain the purpose of __str__ and __repr__ methods in Python.

Ans - __str__ returns a readable, user-friendly string representation, typically for print(obj) or str(obj). __repr__ returns a representation intended to be informative for developers and debugging; ideally it is unambiguous and, where practical, can resemble valid Python code to recreate the object. If __str__ is not defined, Python may fall back to __repr__.


20. What is the significance of the super() function in Python?

Ans - super() provides a way to call methods from the next class in the method resolution order (MRO), commonly a parent class. It is often used in __init__ to initialize inherited parts and in overridden methods to extend rather than replace parent behavior. In multiple inheritance, it supports cooperative method calls.


21. What is the significance of the __del__ method in Python?
Ans - __del__ is a finalizer that may be called when an object is about to be finalized. Its timing is not guaranteed, and it may not run at interpreter shutdown; reference cycles can complicate finalization. It should not be relied on for critical cleanup. Prefer context managers (with) or explicit close() methods for resources.


22. What is the difference between @staticmethod and @classmethod in Python?

Ans - staticmethod defines a method that receives no automatic instance or class argument. classmethod defines a method that receives the class as its first argument (cls), including when called through a subclass. Use static methods for related utilities and class methods when behavior needs the class itself, such as creating instances or updating class-level data.


23. How does polymorphism work in Python with inheritance?

Ans - A parent class can define a common method, and subclasses can override it with their own implementations. When code calls that method on an object, Python dispatches to the implementation for the object's actual type. This is runtime polymorphism. Python also supports duck typing, where compatible behavior matters more than a shared base class.


24. What is method chaining in Python OOP?

Ans - Method chaining is a style in which several method calls are linked in one expression. A method returns self (or another object that supports the next call), allowing expressions such as obj.set_name(...).set_age(...). It can make fluent APIs concise, but methods must return the appropriate object.


25. What is the purpose of the __call__ method in Python?

Ans - Defining __call__ lets an instance be called like a function. For example, if an object has a __call__ method, object_name(arguments) invokes object_name.__call__(arguments). This is useful for callable objects that retain state between calls.

```python
#1. Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog that overrides the speak() method to print "Bark!".

class Animal:
    def speak(self):
        print("The animal makes a sound.")

class Dog(Animal):
    def speak(self):
        print("Bark!")

animal = Animal()
dog = Dog()
animal.speak()
dog.speak()
```

```python
#2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method in both.

from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

print(f"Circle area: {Circle(3).area():.2f}")
print(f"Rectangle area: {Rectangle(4, 5).area():.2f}")
```

```python
#3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car and further derive a class ElectricCar that adds a battery attribute.

class Vehicle:
    def __init__(self, type):
        self.type = type

class Car(Vehicle):
    def __init__(self, type, brand):
        super().__init__(type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, type, brand, battery):
        super().__init__(type, brand)
        self.battery = battery

car = ElectricCar("Electric vehicle", "Example Motors", "75 kWh")
print(car.type, car.brand, car.battery)
```

```python
#4. Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classes Sparrow and Penguin that override the fly() method.

class Bird:
    def fly(self):
        print("Bird flight behavior.")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow can fly.")

class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly.")

for bird in (Sparrow(), Penguin()):
    bird.fly()

```

```python
#5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.

class BankAccount:
    def __init__(self, opening_balance=0):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative.")
        self.__balance = opening_balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive.")
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive.")
        if amount > self.__balance:
            raise ValueError("Insufficient funds.")
        self.__balance -= amount

    def check_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)
print(account.check_balance())
```

```python
#6. Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar and Piano that implement their own version of play().

class Instrument:
    def play(self):
        print("The instrument is being played.")

class Guitar(Instrument):
    def play(self):
        print("Strumming the guitar.")

class Piano(Instrument):
    def play(self):
        print("Playing the piano keys.")

for instrument in (Guitar(), Piano()):
    instrument.play()
```

```python
#7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static method subtract_numbers() to subtract two numbers.

class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

print(MathOperations.add_numbers(10, 5))
print(MathOperations.subtract_numbers(10, 5))
```

```python
#8. Implement a class Person with a class method to count the total number of persons created.

class Person:
    count = 0

    def __init__(self, name):
        self.name = name
        Person.count += 1

    @classmethod
    def total_persons(cls):
        return cls.count

Person("Asha")
Person("Rahul")
Person("Mina")
print(Person.total_persons())

```

```python
#9. Write a class Fraction with attributes numerator and denominator. Override the str method to display the fraction as "numerator/denominator".

class Fraction:
    def __init__(self, numerator, denominator):
        if denominator == 0:
            raise ValueError("Denominator cannot be zero.")
        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

fraction = Fraction(3, 4)
print(fraction)
```

```python
#10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two vectors.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 5)
print(v1 + v2)
```

```python
#12. Implement a class Student with attributes name and grades. Create a method average_grade() to compute the average of the grades.

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        if not self.grades:
            raise ValueError("Cannot calculate an average without grades.")
        return sum(self.grades) / len(self.grades)

student = Student("Sunny", [80, 90, 85])
print(f"{student.name}'s average: {student.average_grade():.2f}")
```

```python
#13. Create a class Rectangle with methods set_dimensions() to set the dimensions and area() to calculate the area.

class Rectangle:
    def __init__(self):
        self.length = 0
        self.width = 0

    def set_dimensions(self, length, width):
        if length < 0 or width < 0:
            raise ValueError("Dimensions cannot be negative.")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

rectangle = Rectangle()
rectangle.set_dimensions(6, 4)
print(rectangle.area())
```

```python
#14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

class Employee:
    def __init__(self, hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, hours_worked, hourly_rate, bonus):
        super().__init__(hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        return super().calculate_salary() + self.bonus

employee = Employee(160, 20)
manager = Manager(160, 30, 500)
print(employee.calculate_salary())
print(manager.calculate_salary())

```

```python
#15. Create a class Product with attributes name, price, and quantity. Implement a method total_price() that calculates the total price of the product.

class Product:
    def __init__(self, name, price, quantity):
        if price < 0 or quantity < 0:
            raise ValueError("Price and quantity cannot be negative.")
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity

product = Product("Notebook", 50, 4)
print(product.total_price())
```

```python
#16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method.

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        return "Moo"

class Sheep(Animal):
    def sound(self):
        return "Baa"

print(Cow().sound())
print(Sheep().sound())
```

```python
#17. Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that returns a formatted string with the book's details.

class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published

    def get_book_info(self):
        return (f"Title: {self.title}, Author: {self.author}, "
                f"Year published: {self.year_published}")

book = Book("The Alchemist", "Paulo Coelho", 1988)
print(book.get_book_info())

```
